```python
import statistics

data = [48, 49, 50, 51, 52]

print(statistics.pvariance(data))  # Population variance
print(statistics.variance(data))   # Sample variance

print(statistics.pstdev(data))    # Population standard deviation
print(statistics.stdev(data))     # Sample standard deviation
```

 1. What is Mean?

Mean is the average of a set of numbers.

Example: Numbers: 10, 20, 30

Mean = (10 + 20 + 30) / 3 = 20
 
 2. What is Median?

Median is the middle value when numbers are arranged in order.

Example: Numbers: 10, 20, 30

Median = 20

 3. What is Mode?

Mode is the value that appears most frequently.

Example: Numbers: 10, 20, 20, 30

Mode = 20

 4. What is Variance?

Variance measures how spread out the numbers are from their mean.

Example: Numbers: 2, 4, 6

Mean = 4. The population variance is 2.67.

 5. What is Standard Deviation?

Standard deviation measures how much the values vary from the mean. It is the square root of variance.

Example: If variance = 4, standard deviation = \\(\sqrt{4}=\\) 2.

 6. What is Probability?

Probability tells us how likely an event is to happen.

Example: When tossing a fair coin, the probability of getting heads is \\(1/2\\), or 50%.
