```python
# create a small dataset
import pandas as pd
data = {'Name': ['Anu', 'Bala', 'Anu', 'Divya'],
        'Age': [20, None, 20, 22],
        'City': ['Chennai', 'Madurai', 'Chennai', 'chennai ']}
df = pd.DataFrame(data)
print(df)
```

```python
# find missing values
import pandas as pd
df = pd.DataFrame({'Age': [20, None, 22, None]})
print(df.isnull().sum())
```

```python
# fill missing values with the median
import pandas as pd
df = pd.DataFrame({'Age': [20, None, 22, 24]})
df['Age'] = df['Age'].fillna(df['Age'].median())
print(df)
```

```python
# remove rows with missing values
import pandas as pd
df = pd.DataFrame({'Age': [20, None, 22]})
print(df.dropna())
```

```python
# find and remove duplicate rows
import pandas as pd
df = pd.DataFrame({'Name': ['Anu', 'Bala', 'Anu'], 'Age': [20, 21, 20]})
print('Duplicates:', df.duplicated().sum())
print(df.drop_duplicates())
```

```python
# change a column to the correct data type
import pandas as pd
df = pd.DataFrame({'Age': ['20', '21', '22']})
df['Age'] = df['Age'].astype(int)
print(df.dtypes)
```

```python
# remove extra spaces and make text consistent
import pandas as pd
df = pd.DataFrame({'City': [' chennai', 'CHENNAI ', 'madurai']})
df['City'] = df['City'].str.strip().str.title()
print(df)
```

```python
# identify outliers using the IQR method
import pandas as pd
values = pd.Series([10, 12, 13, 14, 15, 100])
q1 = values.quantile(0.25)
q3 = values.quantile(0.75)
iqr = q3 - q1
outliers = values[(values < q1 - 1.5 * iqr) | (values > q3 + 1.5 * iqr)]
print('Outliers:', outliers.tolist())
```

```python
# one-hot encoding - turn categories into number columns
import pandas as pd
df = pd.DataFrame({'Color': ['Red', 'Blue', 'Red']})
encoded = pd.get_dummies(df, columns=['Color'], dtype=int)
print(encoded)
```

```python
# simple category-to-number encoding
import pandas as pd
df = pd.DataFrame({'Size': ['Small', 'Medium', 'Large']})
df['Size_Code'] = df['Size'].map({'Small': 1, 'Medium': 2, 'Large': 3})
print(df)
```

## Concepts and Output Explanations

**1. What is Data Cleaning?**  
Data cleaning means fixing missing, repeated or incorrectly formatted data so it can be used correctly.

**2. What are Missing Values?**  
Missing values are empty entries in a dataset. **Example:** An unknown age. **Output:** `isnull().sum()` counts the missing entries.

**3. What is Imputation?**  
Imputation means replacing missing values with a suitable value. **Example:** Fill an empty age with the median. **Output:** The missing age becomes 22.

**4. What is Dropping Missing Values?**  
`dropna()` removes rows containing missing values. **Output:** Only rows with known ages remain.

**5. What are Duplicates?**  
Duplicates are repeated rows. **Example:** Anu, age 20, appears twice. **Output:** `drop_duplicates()` keeps one copy.

**6. What is Data Type Conversion?**  
It changes a value into the type needed for calculations. **Example:** `'20'` (text) becomes `20` (integer). **Output:** The Age column has integer type.

**7. What is Text Standardization?**  
It makes text values consistent. **Example:** `' chennai'` and `'CHENNAI '` become `'Chennai'`. **Output:** Spaces and inconsistent capitalization are removed.

**8. What is an Outlier?**  
An outlier is a value far from most other values. **Example:** 100 in `[10, 12, 13, 14, 15, 100]`. **Output:** The IQR method identifies 100 as an outlier. An outlier should be checked before removing it.

**9. What is One-Hot Encoding?**  
One-hot encoding turns categories into separate 0/1 columns. **Example:** Red and Blue. **Output:** `Color_Red` and `Color_Blue` columns.

**10. What is Category-to-Number Encoding?**  
It replaces categories with numbers. **Example:** Small = 1, Medium = 2, Large = 3. **Output:** A new `Size_Code` column. This ordering is meaningful for sizes, but not for every category.

**Summary:** Data cleaning handles missing values, duplicates, types, text and outliers. Encoding prepares categorical values for numerical analysis.
