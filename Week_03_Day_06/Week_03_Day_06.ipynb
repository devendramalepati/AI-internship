
### 1. What is a vector?

A vector is an ordered collection of numbers used to represent data.

Example: `A = [10, 20, 30]`

### 2. What are vector operations?

Vector operations are mathematical operations performed on vectors, such as addition, subtraction, and scalar multiplication.

Example:\
`A = [1, 2]`, `B = [3, 4]`

`A + B = [4, 6]`

### 3. What is a matrix?

A matrix is a rectangular arrangement of numbers in rows and columns.

Example:

\\[ A=\begin{bmatrix}1&2\\\3&4\end{bmatrix} \\]

This matrix has 2 rows and 2 columns.

### 4. What is the shape of a matrix?

The shape of a matrix tells us the number of rows and columns it contains.

Example: A matrix with 3 rows and 2 columns has shape `(3, 2)` in NumPy.

### 5. What is the dot product?

The dot product multiplies corresponding elements of two vectors and adds the results.

Example:

`A = [1, 2]`, `B = [3, 4]`

Dot product = `(1 × 3) + (2 × 4) = 11`

### 6. What is matrix multiplication?

Matrix multiplication is an operation in which rows of the first matrix are multiplied by columns of the second matrix, and the results are added.

Example:

\\[ A=\begin{bmatrix}1&2\\\3&4\end{bmatrix}, B=\begin{bmatrix}1&0\\\0&1\end{bmatrix} \\]

\\[ AB=\begin{bmatrix}1&2\\\3&4\end{bmatrix} \\]

### 7. What is the difference between `*` and `@` in NumPy?

The `*` operator performs element-wise multiplication, while `@` performs matrix multiplication.

Example: If `A = [1, 2]` and `B = [3, 4]`, then `A * B` gives `[3, 8]`.

### 8. What is a gradient?

A gradient tells us how a function changes when its input changes. It is important in machine learning for adjusting model parameters.

Example: For \\(f(x)=x^2\\), the gradient is \\(2x\\). At \\(x=3\\), the gradient is 6.

### 9. What is gradient descent?

Gradient descent is an optimization method used to reduce a model's error by updating its parameters using gradients.

Example: If a parameter is 5, the learning rate is 0.1, and the gradient is 4, the updated parameter is:

\\[ 5-(0.1\times4)=4.6 \\]

### 10. Why are vectors and matrices important in AI?

Vectors and matrices help represent and process numerical data efficiently. They are widely used in machine learning models and neural networks.

### 11. Why is NumPy used for linear algebra?

NumPy provides functions and operators to perform mathematical operations on vectors and matrices efficiently.

Example: `np.dot(A, B)` calculates the dot product of compatible vectors.

### 12. What is the difference between a vector and a matrix?

A vector is a one-dimensional collection of values, whereas a matrix is a two-dimensional arrangement of values in rows and columns.

### 13. What happens when matrix dimensions are incompatible?

Matrix multiplication fails when the number of columns in the first matrix does not equal the number of rows in the second matrix.

Example: A `(2, 3)` matrix cannot be multiplied by a `(2, 2)` matrix using `@`.

### 14. How are gradients used in machine learning?

Gradients help determine how model parameters should change to reduce the loss or prediction error during training.

### 15. What is the difference between dot product and matrix multiplication?

The dot product of two vectors produces a single number. Matrix multiplication combines two compatible matrices and produces another matrix.
